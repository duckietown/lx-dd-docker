# dts devel Build and Run

`dts devel` is for an authorized Duckiedrone project workflow, not for this learning experience's local Docker practice. Commands that name a physical Duckiedrone can build or load an image, synchronize code, and create containers, so verify the intended target and follow the applicable deployment procedure before running them.

## Use `dts devel` to build, deliver, and run a project

`dts devel` operates on a Duckietown project and invokes Docker for that project. It is for building images and launching project containers, not for replacing the read-only `docker` inspection commands above. A `dts devel` command that names a physical Duckiedrone is a modifying deployment action: it can build or load an image, synchronize code, and create containers.

### Choose the build and delivery machines

`dts devel build` uses Docker Buildx to build the current project's image. Its `-H` or `--machine` option selects the Docker socket or hostname where the image is built. With no `-H`, the build uses the authorized local Docker daemon. Its `-D` or `--destination` option selects where the completed image is delivered. `-D` belongs to `dts devel build`; `dts devel run` has no delivery option.

When the build machine and destination differ, the Duckietown Shell saves the completed image from the build machine and loads it into the destination daemon. Run these only from a valid Duckietown project and only in a deployment workflow you are authorized to carry out:

```bash
# Build the current project on the local base station.
dts devel build
# Build the current project directly on an authorized Duckiedrone.
dts devel build -H DUCKIEDRONE_NAME
# Build locally for an arm64v8 Duckiedrone, then deliver the image to it.
dts devel build -a arm64v8 -D DUCKIEDRONE_NAME
# Build on one authorized Docker host, then deliver to an authorized Duckiedrone.
dts devel build -H BUILD_HOST -a arm64v8 -D DUCKIEDRONE_NAME
```

If the build machine and delivery machine have different CPU architectures, select an image architecture that the delivery machine can run. A build performed on an arm64v8 Duckiedrone can detect that target architecture automatically. A build performed on an amd64 base station normally needs `-a arm64v8` before it is delivered to an arm64v8 Duckiedrone.

A Docker context name and a `dts devel` host argument are different mechanisms. `docker --context duckiedrone-DUCKIEDRONE_NAME` tells the Docker client to use stored local context configuration. `dts devel build -H DUCKIEDRONE_NAME` passes a direct hostname or Docker socket to the Duckietown Shell for that one command. Do not pass a Docker context name to `-H` as though it were a hostname, and do not expect `dts devel` to change the active Docker context.

For a bare `DUCKIEDRONE_NAME`, the current Duckietown Shell resolves the robot hostname and directs Docker to its configured TCP Docker endpoint, which defaults to port `2375`. This is a direct, non-SSH Docker API path, not the SSH-backed context created above. Treat it as a trusted, owner-authorized deployment mechanism only. Do not expose that endpoint, replace a failed remote-context connection with it, append a guessed port to a command, or retry against an unfamiliar Duckiedrone. Stop and review the device's provisioning record and [Duckiedrone DD24 documentation](https://docs.duckietown.com/ente/opmanual-dd24/); in an instructor-led setting, also follow the instructor's procedure.

### Choose the run machine

`dts devel run` creates and runs a container from the current project's image. Its `-H` or `--machine` option selects the Docker socket or hostname where that container runs:

```bash
# Run the project image locally with the normal local project mount.
dts devel run
# Run the project image on an authorized Duckiedrone.
dts devel run -H DUCKIEDRONE_NAME
```

A local run can mount the current project directly into the container. A host bind mount cannot cross from the base station to a remote Duckiedrone. For a normal remote run with mounting enabled, `dts devel run -H DUCKIEDRONE_NAME` automatically synchronizes the project to the remote machine and then runs the container there. `--no-mount` disables both the project mount and that automatic synchronization. The `--sync` option is valid only with a remote `-H` target.

Unlike `docker image ls`, these commands change state on the selected daemon. A build creates an image, a delivery loads an image into another daemon, and a run creates a container. Keep those operations within the authorized project procedure and do not use them to experiment with platform services.
