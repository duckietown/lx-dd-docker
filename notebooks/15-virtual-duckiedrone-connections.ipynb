# Virtual Duckiedrone Connections

This notebook distinguishes a local virtual Duckiedrone shell from access to a physical Duckiedrone.

## Connect to a virtual Duckiedrone

A virtual Duckiedrone runs as a local base-station container. `dts duckiebot virtual connect DUCKIEDRONE_NAME` checks for a local container named `dts-virtual-DUCKIEDRONE_NAME` and opens an interactive shell inside it with `docker exec`. It does not create a remote Docker context.

Before using it, use the checks in Notebook 11 to identify the authorized local Docker context and confirm that no remote Docker environment variable is set. In the commands below, replace `LOCAL_DOCKER_CONTEXT` with that context name and `DUCKIEDRONE_NAME` with the name assigned by the approved virtual Duckiedrone setup procedure; do not type either placeholder literally. You can then check whether the virtual container is present without entering it:

```bash
docker --context LOCAL_DOCKER_CONTEXT container ls --all \
  --filter 'name=^/dts-virtual-DUCKIEDRONE_NAME$'
dts duckiebot virtual connect DUCKIEDRONE_NAME
```

Once inside the virtual Duckiedrone shell, `docker image ls` and `docker container ls` inspect that virtual Duckiedrone's own Docker daemon. This is analogous to using SSH to access a physical Duckiedrone and running Docker commands there. If the virtual-connect command reports that no virtual robot is running, use the approved virtual Duckiedrone setup procedure. Do not switch to a physical Duckiedrone context to make the command work.

## Choose the right connection path

| Goal | Connection path | Docker target |
| --- | --- | --- |
| Complete this learning experience's hands-on exercises | Local Docker context | Learner-owned `lx-docker-*` resources on the base station |
| Inspect a physical Duckiedrone through its own shell | SSH, then a read-only Docker command | Platform daemon on the physical Duckiedrone |
| Inspect a physical Duckiedrone without opening SSH | Explicit authorized Docker context | Platform daemon on the physical Duckiedrone |
| Open a shell in a running virtual Duckiedrone | `dts duckiebot virtual connect DUCKIEDRONE_NAME` | Local `dts-virtual-DUCKIEDRONE_NAME` container |

In all cases, confirm the target before acting. A context name is only local client configuration; it does not make a remote Docker daemon safe, reachable, or authorized. A wrong Docker context or an unauthorized `dts` deployment target can send ordinary-looking commands to a physical Duckiedrone and unintentionally disrupt platform, sensor, or flight-support services.
