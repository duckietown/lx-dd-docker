# Virtual Duckiedrone Connections

This module distinguishes a local virtual Duckiedrone shell from Secure Shell (SSH) access to a physical Duckiedrone. Confirm the intended target before acting, and use the approved virtual-device setup procedure rather than switching to a physical target when a virtual connection is unavailable.

## Connect to a virtual Duckiedrone

A virtual Duckiedrone runs as a local base-station container. `dts duckiebot virtual connect ROBOT_NAME` checks for a local container named `dts-virtual-ROBOT_NAME` and opens an interactive shell inside it with `docker exec`. It does not create a remote Docker context.

Before using it, confirm that the local Docker context is active and that no remote Docker environment variable is set. You can check whether the virtual container is present without entering it:

```bash
docker --context default container ls --all \
  --filter 'name=^/dts-virtual-ROBOT_NAME$'
dts duckiebot virtual connect ROBOT_NAME
```

Once inside the virtual Duckiedrone shell, `docker image ls` and `docker container ls` inspect that virtual Duckiedrone's own Docker daemon. This is analogous to using SSH to access a physical Duckiedrone and running Docker commands there. If the virtual-connect command reports that no virtual robot is running, use the approved virtual-Duckiedrone setup procedure. Do not switch to a physical Duckiedrone context to make the command work.

## Choose the right connection path

| Goal | Connection path | Docker target |
| --- | --- | --- |
| Complete this learning experience's hands-on exercises | Local Docker context | Learner-owned `lx-docker-*` resources on the base station |
| Inspect a physical Duckiedrone through its own shell | SSH, then a read-only Docker command | Platform daemon on the physical Duckiedrone |
| Inspect a physical Duckiedrone without opening SSH | Explicit authorized Docker context | Platform daemon on the physical Duckiedrone |
| Build and deliver a project image in an authorized later learning experience | `dts devel build` with explicit `-H` and, when needed, `-D` | Selected builder and delivery Docker daemons |
| Build and launch an agent in a later learning experience | `dts code build -R ROBOT_NAME`, then `dts code workbench -R ROBOT_NAME` | Platform daemon on the physical Duckiedrone |
| Open a shell in a running virtual Duckiedrone | `dts duckiebot virtual connect ROBOT_NAME` | Local `dts-virtual-ROBOT_NAME` container |

In all cases, confirm the target before acting. A context name is only local client configuration; it does not make a remote Docker daemon safe, reachable, or authorized.
