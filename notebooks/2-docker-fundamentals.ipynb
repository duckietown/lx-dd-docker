## Containers are not virtual machines

A __container__ is an isolated process with the files and configuration it needs to run. Containers on the same computer share the host Linux kernel. A __virtual machine__ has its own guest operating system and kernel, which usually needs more memory, storage, and startup time.

| | Container | Virtual machine |
| --- | --- | --- |
| Kernel | Shares the host kernel | Runs a guest kernel |
| Startup and resource cost | Usually lower | Usually higher |
| Isolation | Process, filesystem, and resource boundaries | A full guest operating system |

A container is not a complete computer and it is not an absolute security boundary. Its access depends on the permissions, files, devices, network settings, and resources given to it when it starts. For example, mounting a host directory lets a container read or change that directory according to the mount permissions.

## Images, layers, and containers

An __image__ is a read-only package used to create containers. It can contain an operating-system user space, libraries, applications, and default settings. An image is built from __layers__, ordered filesystem changes. Docker can reuse unchanged layers when building or downloading related images.

A __container__ is a running or stopped instance created from an image. It begins with the image's files, but files changed inside a container belong to that container's writable layer. Those changes are removed when the container is removed unless you deliberately use a volume or bind mount.

```text
Dockerfile -> image layers -> image -> container -> another container
```

Running the same image twice creates two separate containers. Stopping a container does not remove it. The `--rm` option used later removes a temporary container automatically after it exits.
