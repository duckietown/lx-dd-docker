# Docker Contexts and Local Targets

A Docker context tells the Docker client which daemon receives a command. This notebook compares Docker commands run from a Duckiedrone shell with commands run through an explicit Docker context on the base station.

## Inspect the current base-station target

Start every base-station Docker session by checking the active context and the available context names:

```bash
docker context show
docker context ls
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
```

For the local Docker exercises, `docker context show` must identify the authorized local Docker host. On many Linux systems this context is named `default`, but use the name shown on your base station. The `DOCKER_HOST` and `DOCKER_CONTEXT` environment variables can change the endpoint that the Docker client uses, so check them whenever the result is surprising.

`docker context ls` can reveal endpoint names and connection details from other work on a shared computer. Treat that information as private operational data and do not copy it into activity reports.

## Make the base-station target explicit

The `--context` option applies to one Docker command without changing the context used by later commands. For a local check, replace `default` with your authorized local context name when necessary:

```bash
docker --context default version
docker context show
```

`docker version` contacts the selected daemon and reports the client and server versions. The second command should still show the context that was active before the explicit check. Prefer `--context NAME` for a short, authorized verification instead of `docker context use NAME`, which changes the default target for later unqualified Docker commands.

Do not run `docker stop`, `docker rm`, `docker run`, image removal, or cleanup commands until `docker context show` and the intended target agree. A wrong context can make those commands affect the base station, a physical Duckiedrone, or another authorized host unexpectedly.
