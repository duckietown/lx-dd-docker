# Docker Contexts and Local Targets

A Docker context tells the Docker client which daemon receives a command. This notebook compares Docker commands run from a Duckiedrone shell with commands run through an explicit Docker context on the base station.

## Inspect the current base-station target

Start every base-station Docker session by checking the selected context, the available context names, and environment variables that can alter the target:

```bash
docker context show
docker context ls
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
```

`docker context show` prints the selected context name, while `docker context ls` lists the endpoint configured for each context. Neither result alone proves the daemon that an unqualified Docker command will reach: `DOCKER_CONTEXT` can select a named context and `DOCKER_HOST` can select a daemon directly. For the local Docker exercises, use the authorized local context named by your base-station setup. On many Linux systems it is named `default`, but use the name shown on your base station.

`docker context ls` can reveal endpoint names and connection details from other work on a shared computer. Treat that information as private operational data and do not copy it into activity reports.

## Make the base-station target explicit

The `--context` option selects a named context for one Docker command without changing the context used by later commands. Assign `LOCAL_CONTEXT` to your authorized local context; `default` below is only an example:

```bash
LOCAL_CONTEXT=default  # Replace default with your authorized local context name.
docker --context "$LOCAL_CONTEXT" version
docker context show
```

`docker --context "$LOCAL_CONTEXT" version` contacts the daemon configured for that named context and reports the client and server versions without changing resources. The second command still shows the context that unqualified Docker commands would use. Prefer an explicit `--context NAME` for a short, authorized verification instead of `docker context use NAME`, which changes the default selection for later unqualified Docker commands.

Do not run `docker stop`, `docker rm`, `docker run`, image removal, or cleanup commands until `LOCAL_CONTEXT` names an authorized local context, `docker --context "$LOCAL_CONTEXT" version` has contacted the expected daemon, and you understand any nonempty `DOCKER_HOST` or `DOCKER_CONTEXT` setting. A wrong target can make those commands affect the base station, a physical Duckiedrone, or another authorized host unexpectedly.
