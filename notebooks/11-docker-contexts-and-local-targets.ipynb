# Docker Contexts and Local Targets

A Docker context tells the Docker client which daemon receives a command. This notebook compares Docker commands run from a Duckiedrone shell with commands run through an explicit Docker context on the base station.

## Inspect the current base-station target

[Notebook 2](./2-docker-containers-images-and-safe-local-practice.ipynb) introduces the target check required before commands that modify Docker resources. For the local Docker exercises, use the authorized local context named by your base-station setup. On many Linux systems it is named `default`, but use the name shown on your base station. Start by checking the selected context:

```bash
docker context show
```

Then list the available named contexts and their configured endpoints. This output can reveal endpoint names and connection details from other work on a shared computer, so treat it as private operational data and do not copy it into activity reports:

```bash
docker context ls
```

`DOCKER_HOST` and `DOCKER_CONTEXT` can override the selected context. Check their values before using a modifying command:

```bash
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
```

## Make the base-station target explicit

The `--context` option selects a named context for one Docker command without changing the context used by later commands. Prefer an explicit `--context NAME` for a short, authorized verification instead of `docker context use NAME`, which changes the default selection for later unqualified Docker commands. Assign `LOCAL_CONTEXT` to your authorized local context. The command below uses `default` only as an example; replace it with your authorized local context name:

```bash
LOCAL_CONTEXT=default
```

Contact the daemon configured for that named context and report its client and server versions without changing resources:

```bash
docker --context "$LOCAL_CONTEXT" version
```

That explicit command does not change the context used by later unqualified Docker commands. Confirm the unqualified selection remains unchanged:

```bash
docker context show
```

Do not run `docker stop`, `docker rm`, `docker run`, image removal, or cleanup commands until `LOCAL_CONTEXT` names an authorized local context, `docker --context "$LOCAL_CONTEXT" version` has contacted the expected daemon, and you understand any nonempty `DOCKER_HOST` or `DOCKER_CONTEXT` setting. A wrong target can make those commands affect the base station, a Duckiedrone, or another authorized host unexpectedly.
