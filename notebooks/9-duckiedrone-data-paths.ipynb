# Duckiedrone Data Paths

A camera image or flight-controller reading passes through several services before an application can use it. Knowing that route helps you identify which driver or bridge to inspect when data is missing. We will connect the optional stacks to their capabilities, then trace sensor and flight-controller data through the Duckiedrone.

## Optional services

The `duckiedrone/extra_sensors` stack adds drivers and ROS 2 bridges for the front, left, right, and top ToF sensors. The drivers publish readings through DTPS, and the bridges make them available to ROS 2 programs. These services start when you request the optional stack. They use Docker's `unless-stopped` restart policy, so they restart after a reboot unless explicitly stopped or the stack is taken down.

The `duckiedrone/extra_tools` stack adds the `ros2-foxglove-bridge` service for viewing and recording ROS 2 data. It is also started on request.

The separate `ros2-core/duckiedrone` stack defines the `altitude` service and has its own startup instructions.

Reading a physical sensor requires its driver to access the sensor hardware. A ROS 2 bridge reads the data published by the driver, so it needs different connections. The stack's Compose file records the connections for each container. A local practice container needs only the files and connections used by its exercise.

## Duckiedrone data paths

Sensor drivers run in containers and publish sensor data through `dtps`. ROS 2 bridge containers read that data and republish it for ROS 2 programs. The path is summarized in [Figure 1](#figure-1).

  <pre style="display:inline-block; margin:0; text-align:left;">
    camera or bottom ToF sensor
      |
      v
    driver container
      |
      v
    dtps
      |
      v
    ROS 2 bridge container
      |
      v
    Duckiedrone service
  </pre>

On a physical Duckiedrone, PX4 flight-control firmware runs on the flight-controller board. The `ros2-mavros` container exchanges [Micro Air Vehicle Link (MAVLink)](https://mavlink.io/) messages with that controller and makes its state and telemetry available to ROS 2 programs. On a virtual Duckiedrone, `dt-px4` runs PX4 in a container on the base station. The browser-based Dashboard receives ROS 2 data through `ros2-rosbridge-websocket`.
