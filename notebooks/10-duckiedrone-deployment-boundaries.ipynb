# Duckiedrone Deployment Boundaries

The Duckiedrone's sensor and communication services work together, so starting or updating them needs to preserve their settings and connections. `dts` provides commands for this work. We will compare them with local Docker practice, identifying which computer each command affects before changing its images or containers.

## Deployment boundary

The Duckiedrone manual's [software update procedure](https://docs.duckietown.com/ente/opmanual-dd24/40-duckiedrone-handling/10-software-update.html) uses `dts duckiebot update DUCKIEDRONE_NAME` from a base-station terminal. It downloads the Duckiedrone's images and starts the default stacks.

A stack is a group of containers whose settings are described together in a Docker Compose file. The standard update starts `robot/basics`, `duckietown/duckiedrone`, and `ros2/duckiedrone`. Follow [Preparing the software stack](https://docs.duckietown.com/ente/opmanual-dd24/40-duckiedrone-handling/environment-setup.html) to check the expected services after updating.

For optional sensor services, the manual's [container reference](https://docs.duckietown.com/ente/opmanual-dd24/60-software-architecture/containers-running-on-the-duckiedrone.html#optional-sensors) uses `dts stack up -H DUCKIEDRONE_NAME -d duckiedrone/extra_sensors`. Here `-H` selects the Duckiedrone and `-d` runs the stack in the background. Starting this optional stack is separate from updating the default stacks.

The command's target determines which computer's images and containers it uses. The update and stack examples select the Duckiedrone by name; Docker commands use the configured daemon connection. Select the base station's local Docker context for the local exercises.
