# Duckiedrone Deployment Boundaries

This module separates authorized Duckiedrone deployment work from local Docker practice. Do not use a physical Duckiedrone or its platform containers for experimentation; follow the device owner's approved procedure and any instructor guidance.

## Deployment boundary

A base-station Duckietown Shell command such as `dts stack` with `-H ROBOT_NAME` manages a named stack through the Duckiedrone's remote Docker endpoint. That is a platform deployment workflow, not the same operation as running `docker` against the authorized local daemon for this learning experience (LX).

The Docker Application Programming Interface (API) is commonly associated with Transmission Control Protocol (TCP) ports `2375` and `2376`, but those port numbers do not reveal how a particular remote endpoint is configured or secured. Do not set `DOCKER_HOST` to reach a Duckiedrone, expose a Docker API, or reproduce an endpoint configuration from this notebook. Use only documented Duckietown commands when an authorized operating procedure calls for them.

## Keep the roles separate

| Task | Correct workflow |
| --- | --- |
| Learn image, container, port, and volume basics | Use the local `lx-docker-*` resources in Notebooks 2-7 |
| Deploy or update Duckiedrone platform services | Use the authorized Duckietown operating procedure |
| Modify or remove a platform container | Do not do this as a Docker practice exercise |

Keeping those roles separate protects the Duckiedrone's sensor and flight-support services while still allowing safe hands-on Docker practice on the base station.
