# Duckiedrone Deployment Boundaries

This notebook distinguishes Duckiedrone deployment work from local Docker practice.

## Deployment boundary

A base-station command from the Duckietown Shell, such as `dts stack up -H DUCKIEDRONE_NAME STACK_NAME`, starts a named stack through the Duckiedrone's remote Docker endpoint. `STACK_NAME` is a stack named by the authorized operating procedure. That is a platform deployment workflow, not the same operation as running `docker` against the authorized local daemon for this LX.

The Docker API is commonly associated with TCP ports `2375` and `2376`, but those port numbers do not reveal how a particular remote endpoint is configured or secured. Do not set `DOCKER_HOST` to reach a Duckiedrone, expose a Docker API, or reproduce an endpoint configuration from this notebook. Use only documented Duckietown commands when an authorized operating procedure calls for them.
