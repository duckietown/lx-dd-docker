# Duckiedrone Platform Data Paths

This notebook identifies the platform-specific sensor and flight-support paths on a physical Duckiedrone.

## Platform extensions

The `duckiedrone/extra_sensors` stack adds the front, left, right, and top time-of-flight (ToF) sensor-driver containers and supporting bridge containers. These services depend on the standard platform stack and exchange sensor data through the shared DTPS socket. Its services use the `unless-stopped` restart policy, so they restart after a reboot until they are explicitly stopped or the named stack is taken down.

The `duckiedrone/extra_tools` stack adds an optional visualization service. It relies on standard platform services, but it is not needed to fly the Duckiedrone or use the Dashboard. The optional manifests use separate Compose project names so that lifecycle operations remain scoped to their named stack.

A separately managed altitude stack contains the `altitude` service. It provides altitude functionality but is not selected by the standard update. Its current manifest uses host networking, a platform-specific environment setting, the `/data` directory, the Avahi socket, and an `unless-stopped` restart policy. Those settings connect it to the deployed Duckiedrone platform; they are not a template for a local practice container.

Hardware-facing services in the current optional sensor manifest intentionally use host networking, `privileged` access, and selected host mounts. Its supporting services share platform sockets. These are reviewed platform-specific exceptions, not Docker defaults and not examples to copy into `lx-docker-*` practice containers.

## Platform data paths

Sensor drivers run in containers and publish sensor data through `dtps`. Supporting bridge containers make that data available to other platform services. The path is summarized in [Figure 1](#figure-1).

<figure id="figure-1" class="lx-figure">
  <pre style="display:inline-block; margin:0; text-align:left;">
    camera or bottom ToF sensor
      |
      v
    driver container
      |
      v
    dtps
      |
      v
    supporting bridge container
      |
      v
    platform service
  </pre>
  <figcaption>
    Figure 1: Sensor data path through a driver, DTPS, and bridge service.
  </figcaption>
</figure>

Flight-control firmware runs on the physical flight controller, not in a container on a physical Duckiedrone. A platform bridge container exchanges state, telemetry, and authorized control-related traffic between it and other platform services. The browser-based Dashboard reaches platform services through a web-service bridge. A separate simulation container is used only for a virtual Duckiedrone.
