# Duckiedrone Platform Data Paths

This conceptual module traces the platform-specific sensor, Robot Operating System 2 (ROS 2), and flight-control paths on a physical Duckiedrone. It does not ask you to start, stop, reconfigure, or inspect a physical Duckiedrone; use documented operating procedures only when you are authorized to do so.

## Optional platform extensions

The `duckiedrone/extra_sensors` stack adds the front, left, right, and top time-of-flight (ToF) sensor drivers and one ROS 2 bridge for each sensor. The bridge services wait for `zenoh-router` from the standard `ros2/duckiedrone` stack, while the driver and bridge services exchange sensor data through the shared Duckietown Postal Service (DTPS) socket. Its services use the `unless-stopped` restart policy, so they restart after a reboot until they are explicitly stopped or the named stack is taken down.

The `duckiedrone/extra_tools` stack adds `ros2-foxglove-bridge` for inspecting and visualizing the Duckiedrone's ROS 2 data. It also relies on the standard `zenoh-router` service, but it is not needed to fly the Duckiedrone or use the Dashboard. The optional manifests use separate Compose project names so that lifecycle operations remain scoped to their named stack.

The separate `ros2-core/duckiedrone` stack contains the `altitude` service. It provides altitude functionality but is not selected by the standard update. Its current manifest uses host networking, `ROS_DOMAIN_ID=42`, the `/data` directory, the Avahi socket, and an `unless-stopped` restart policy. Those settings connect it to the deployed Duckiedrone platform; they are not a template for a local practice container.

Hardware-facing services in the current optional sensor manifest intentionally use host networking, `privileged` access, and selected host mounts. Its ROS 2 bridge services share the DTPS and Avahi sockets. These are reviewed platform-specific exceptions, not Docker defaults and not examples to copy into `lx-docker-*` practice containers.

## Sensor, ROS 2, and flight-control paths

Sensor drivers run in containers and publish sensor data through `dtps`. Companion ROS 2 bridge containers republish the data as ROS 2 topics. `zenoh-router` connects the ROS 2 containers.

```text
camera or bottom ToF sensor
  -> driver container
  -> dtps
  -> ROS 2 bridge container
  -> ROS 2 topics
```

The open-source PX4 autopilot firmware runs on the physical flight controller, not in a container on a physical Duckiedrone. The `ros2-mavros` container exchanges state, telemetry, commands, and arming or mode-control requests between PX4 and ROS 2. The browser-based Dashboard reaches ROS 2 through `ros2-rosbridge-websocket`. A `dt-px4` container is used only for a virtual Duckiedrone simulation.
