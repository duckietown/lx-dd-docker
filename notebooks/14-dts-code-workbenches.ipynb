# dts code Workbenches

This notebook introduces the higher-level workbench workflow used by Duckiedrone projects.

## Use `dts code` for project workbenches

Projects that provide a recipe and launcher use higher-level `dts code` commands. Run the standard sequence from the project root:

Use this target workflow only within an authorized Duckiedrone deployment procedure; it is not a local Docker practice workflow.

```bash
dts code build -R DUCKIEDRONE_NAME
dts code workbench -R DUCKIEDRONE_NAME
```

`dts code build -R DUCKIEDRONE_NAME` prepares the project's recipe and delegates the image build to `dts devel build`. In the usual form above, `-R` names the Duckiedrone and becomes the build machine unless the command is explicitly run with `--local`. The image is therefore built for the target workflow rather than as a generic local Docker exercise.

`dts code workbench -R DUCKIEDRONE_NAME` prepares the same recipe, invokes `dts devel run` for the Duckiedrone, and starts the project's default launcher in a workbench container. That launcher starts the project's standard process in the workbench container. After changing project code, follow the project's instructions to build and start the workbench again.

A project can open a second terminal into the already-running workbench:

```bash
dts code workbench -R DUCKIEDRONE_NAME --shell
```

`--shell` attaches to the existing workbench container; it does not build an image or start a workbench. Start `dts code workbench -R DUCKIEDRONE_NAME` first and leave it running. The attached shell provides access to the configured workbench environment.

Use the `-R DUCKIEDRONE_NAME` form supplied by the project. In `dts code workbench`, `-R` names the Duckiedrone that the project connects to and normally determines where the workbench runs; an explicit `-H` is an advanced override for an authorized workflow where the Docker machine differs from the robot name. `--local` requests a local build or workbench rather than a robot-hosted one. Neither option selects or changes a Docker context, and neither makes a physical Duckiedrone safe to modify without authorization.
