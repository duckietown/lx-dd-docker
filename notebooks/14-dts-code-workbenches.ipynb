# dts code Workbenches

This module introduces the higher-level workbench workflow used by later Duckiedrone learning experiences. It is not a replacement for local Docker practice, and a physical Duckiedrone remains a target only within an authorized deployment procedure.

## Use `dts code` for the later learning-experience workflow

The altitude proportional-integral-derivative (PID), Inertial Measurement Unit (IMU), and Time-of-Flight (ToF) learning experiences use higher-level `dts code` commands because their projects provide a recipe and launcher. The standard sequence is run from the root of the later learning experience:

```bash
dts code build -R ROBOT_NAME
dts code workbench -R ROBOT_NAME
```

`dts code build -R ROBOT_NAME` prepares the project's recipe and delegates the image build to `dts devel build`. In the usual form above, `-R` names the Duckiedrone and becomes the build machine unless the command is explicitly run with `--local`. The image is therefore built for the target workflow rather than as a generic local Docker exercise.

`dts code workbench -R ROBOT_NAME` prepares the same recipe, invokes `dts devel run` for the Duckiedrone, and starts the project's default launcher in a workbench container. That launcher is why later lessons can start a Robot Operating System 2 (ROS 2) node without manually composing a `docker run` command. After changing project code, follow the later learning experience instructions to build and start the workbench again.

A later learning experience can open a second terminal into the already-running workbench:

```bash
dts code workbench -R ROBOT_NAME --shell
```

`--shell` attaches to the existing workbench container; it does not build an image or start a workbench. Start `dts code workbench -R ROBOT_NAME` first and leave it running. The attached shell is where later lessons can run ROS 2 diagnostic commands in the workbench environment.

Use the `-R ROBOT_NAME` form supplied by the later learning experience. In `dts code workbench`, `-R` names the Duckiedrone that the project connects to and normally determines where the workbench runs; an explicit `-H` is an advanced override for an authorized workflow where the Docker machine differs from the robot name. `--local` requests a local build or workbench rather than a robot-hosted one. Neither option selects or changes a Docker context, and neither makes a physical Duckiedrone safe to modify without authorization.
