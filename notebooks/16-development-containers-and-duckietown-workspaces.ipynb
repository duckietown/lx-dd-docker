# Development Containers and Duckietown Workspaces

This notebook applies Docker concepts to development containers and [Duckietown Workspaces](https://github.com/duckietown/workspace).

## Relate terminal environments to Docker

A local terminal runs directly on your development computer. It can use a locally installed Docker Engine or Docker Desktop when your account has permission to access it.

A development container is an isolated development environment, often opened by Visual Studio Code (VS Code). It can include tools and LX files without changing the rest of the computer. A development container might have its own Docker daemon, connect to another daemon through an explicitly configured socket or context, or have no Docker access at all. Use the client, daemon, and context model from the earlier notebooks to determine which arrangement applies.

The Duckietown Workspace is an optional development container project maintained by Duckietown. Its current documentation describes `dts`, Docker-in-Docker for virtual robots, Python and Jupyter tooling, and a browser-accessible desktop. It is useful when your workflow needs those tools, but it is not required for the concepts or local Docker exercises in this LX.

## Separate the shell from the Docker daemon

Apply the client-daemon model from [Notebook 2](./2-docker-containers-images-and-safe-local-practice.ipynb): a development container can expose its own filesystem while its Docker client targets a daemon elsewhere. A local terminal can instead target a local Docker Engine or Docker Desktop installation.

## Verify the current environment

Run these read-only checks from the terminal you plan to use. First, identify the shell and its current files:

```bash
hostname
pwd
```

Check whether the Docker client is available:

```bash
command -v docker
```

Use the target checks from [Notebook 11](./11-docker-contexts-and-local-targets.ipynb) before a modifying Docker command:

```bash
docker context show
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
docker version
```

Check whether `dts` is available. The `|| printf ...` fallback prints a message only when it is not:

```bash
command -v dts || printf '%s\n' 'dts is not installed in this terminal.'
```

If Docker is unavailable or points somewhere unexpected, stop. Follow the setup or authorization guidance in [Notebook 2](./2-docker-containers-images-and-safe-local-practice.ipynb) and [Notebook 11](./11-docker-contexts-and-local-targets.ipynb) rather than changing access settings merely to make a command succeed.
