# Development Containers and Duckietown Workspaces

This notebook applies Docker concepts to development containers and Duckietown Workspaces.

## Relate terminal environments to Docker

A local terminal runs directly on your development computer. It can use a locally installed Docker Engine or Docker Desktop when your account has permission to access it.

A development container is an isolated development environment, often opened by Visual Studio Code (VS Code). It can include tools and LX files without changing the rest of the computer. A development container might have its own Docker daemon, connect to another daemon through an explicitly configured socket or context, or have no Docker access at all. Use the client, daemon, and context model from the earlier notebooks to determine which arrangement applies.

The Duckietown Workspace is an optional dev-container project maintained by Duckietown. Its current documentation describes the Duckietown Shell (`dts`), Docker-in-Docker for virtual robots, Python and Jupyter tooling, and a browser-accessible desktop. It is useful when your workflow needs those tools, but it is not required for the concepts or local Docker exercises in this LX.

## Separate the shell from the Docker daemon

Your terminal, filesystem, and Docker daemon can be in different places. For example, a command typed in a development container can inspect files in that container while its Docker client connects to a daemon elsewhere. A command typed in a local terminal can instead target a local Docker Engine or Docker Desktop installation.

Use the command target, not the terminal label, to decide what a Docker command can affect. A container hostname, a prompt theme, or the presence of VS Code does not prove which daemon receives a Docker request.

## Verify the current environment

Before using a modifying Docker command, run these read-only checks from the terminal you plan to use:

```bash
hostname
pwd
command -v docker
docker context show
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
docker version
command -v dts || printf '%s\n' 'The Duckietown Shell is not installed in this terminal.'
```

`hostname` and `pwd` identify the shell and its current files. `command -v docker` checks whether the Docker client is available. `docker context show` reports the selected context name, while `DOCKER_HOST` and `DOCKER_CONTEXT` can change the target. `docker version` contacts the daemon selected by that configuration but does not create, change, or remove resources. Before a modifying command, confirm that any nonempty override is expected and that the contacted daemon is an authorized target.

If Docker is unavailable or points somewhere unexpected, stop before continuing. Follow the computer owner's or Docker administrator's setup instructions, or inspect an environment you own. Do not add `sudo`, expose a Docker socket, set `DOCKER_HOST` to a guessed address, or switch Docker contexts just to make a command succeed.

## Apply the model

Use these checks whenever you change terminal environments. If the Docker target is unfamiliar, revisit Docker Contexts and Local Targets before using a modifying command. The earlier notebooks explain Docker contexts and authorized Duckiedrone connections separately. A Duckietown Workspace can support virtual-device workflows, but it does not make a physical Duckiedrone Docker daemon an appropriate target for Docker practice.
