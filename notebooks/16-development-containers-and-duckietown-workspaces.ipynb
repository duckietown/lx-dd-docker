# Development Containers and Duckietown Workspaces

Changing terminals can change the available tools, project files, and Docker connection. Understanding those changes helps explain why a command works in one environment or shows a different inventory in another. Development containers package a project's tools; [Duckietown Workspaces](https://github.com/duckietown/workspace) provide one such environment. We will compare terminal types, then check the shell, tools, and daemon separately.

## Relate terminal environments to Docker

A local terminal runs directly on your development computer and can use its Docker Engine or Docker Desktop installation.

A development container is an isolated development environment, often opened by Visual Studio Code (VS Code). It can supply tools and LX files while keeping their dependencies separate from the rest of the computer. Its Docker client may connect to a daemon inside the container or to another daemon through a configured socket or context. The development container's configuration determines whether Docker is available and which daemon its client can reach.

The Duckietown Workspace is an optional development-container project maintained by Duckietown. Its documentation describes `dts`, Docker-in-Docker for virtual Duckiedrones, Python and Jupyter tooling, and a browser-accessible desktop. Docker-in-Docker means that a Docker daemon runs inside a container, managing its own images and containers. The Workspace is useful when a project needs that combination of tools. The local Docker exercises also work from a prepared base-station terminal.

## Separate the shell from the Docker daemon

Apply the model from [Notebook 2](./2-docker-containers-images-and-safe-local-practice.ipynb): the shell's filesystem and the Docker client's target are separate concerns. A development container can show its own project files while sending Docker commands elsewhere. In particular, a bind-mount source refers to files on the daemon's host, which explains why a path visible to the shell may be unavailable to a container started on another host.

## Verify the current environment

Run these checks from the terminal you plan to use. Identify the shell's host and current directory so you know where ordinary file commands operate:

```bash
hostname
pwd
```

Locate the Docker client available to that shell:

```bash
command -v docker
```

Use the target checks from [Notebook 11](./11-docker-contexts-and-local-targets.ipynb) to inspect its connection settings and contact the daemon:

```bash
docker context show
printf 'DOCKER_HOST=%s\n' "${DOCKER_HOST:-<unset>}"
printf 'DOCKER_CONTEXT=%s\n' "${DOCKER_CONTEXT:-<unset>}"
docker version
```

The context and override values explain the target selection. A server section in `docker version` confirms that the selected daemon responded, establishing connectivity for later Docker operations.

Locate `dts` to check whether this terminal can start Duckietown workflows. The fallback prints a message when the tool is absent:

```bash
command -v dts || printf '%s\n' 'dts is not installed in this terminal.'
```

If a tool is missing, return to the environment's setup instructions or choose the prepared base-station terminal. If the Docker target is unexpected, use [Notebook 11](./11-docker-contexts-and-local-targets.ipynb) to identify the configured endpoint before creating exercise resources.
